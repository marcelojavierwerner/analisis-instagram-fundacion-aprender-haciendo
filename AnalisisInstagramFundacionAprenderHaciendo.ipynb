```python
# Clasificacion de Temas
def clasificar_tema(texto):
    texto = str(texto).lower()

    flora = [
        "espinillo", "tala", "sombra de toro", "algarrobo",
        "pasionaria", "árbol", "arbol", "planta", "flora",
        "bosque", "vegetación", "vegetacion"
    ]

    fauna = [
        "ave", "aves", "pájaro", "pajaros", "pájaros",
        "corzuela", "fauna", "animal", "avistaje",
        "reconocimiento de aves"
    ]

    formacion = [
        "capacitación", "capacitacion", "curso", "taller",
        "formación", "formacion", "aprender", "aprendizaje",
        "estudiantes", "unc", "preinscripción", "preinscripcion",
        "inscripción", "inscripcion"
    ]

    actividades = [
        "caminata", "trekking", "jornada", "encuentro",
        "experiencia", "actividad", "senderismo"
    ]

    institucional = [
        "fundación", "fundacion", "gracias", "compromiso",
        "comunidad", "despedimos", "contacto"
    ]

    if any(palabra in texto for palabra in flora):
        return "Flora y naturaleza"

    elif any(palabra in texto for palabra in fauna):
        return "Fauna"

    elif any(palabra in texto for palabra in formacion):
        return "Formación y aprendizaje"

    elif any(palabra in texto for palabra in actividades):
        return "Actividades y experiencias"

    elif any(palabra in texto for palabra in institucional):
        return "Institucional / comunidad"

    else:
        return "Otros"
```

```python
df["tema"] = df["caption"].apply(clasificar_tema)
```

```python
alto_formato = df.groupby("isVideo").agg(
    publicaciones=("id", "count"),
    publicaciones_alto=(
        "nivel_interaccion",
        lambda x: (x == "Alto").sum()
    )
)

alto_formato["porcentaje_alto"] = (
    alto_formato["publicaciones_alto"] /
    alto_formato["publicaciones"] * 100
)

alto_formato["formato"] = alto_formato.index.map({
    False: "Imagen",
    True: "Video"
})

plt.figure(figsize=(8, 5))

plt.bar(
    alto_formato["formato"],
    alto_formato["porcentaje_alto"]
)

plt.xlabel("Formato")
plt.ylabel("Publicaciones Alto (%)")
plt.title("Porcentaje de publicaciones de alto rendimiento según formato")

plt.show()
# Grafico Alto Rendimiento por Formato
```

```python
alto_dia = df.groupby("dia_semana").agg(
    publicaciones=("id", "count"),
    publicaciones_alto=(
        "nivel_interaccion",
        lambda x: (x == "Alto").sum()
    )
)

alto_dia["porcentaje_alto"] = (
    alto_dia["publicaciones_alto"] /
    alto_dia["publicaciones"] * 100
)

alto_dia = alto_dia.sort_values("porcentaje_alto")

plt.figure(figsize=(10, 6))

plt.barh(
    alto_dia.index,
    alto_dia["porcentaje_alto"]
)

plt.xlabel("Publicaciones Alto (%)")
plt.ylabel("Día de la semana")
plt.title("Porcentaje de publicaciones de alto rendimiento por día")

plt.show()
# Porcentaje Alto Rendimiento por Dia
```
